## 1. Sintaxis de Python y comparación con Java

En Java las variables suelen declararse con tipos explícitos y los bloques se delimitan con llaves (`{ }`). En Python el tipo se asocia al valor en tiempo de ejecución y los bloques se delimitan mediante **indentación**.

| Concepto | Java | Python |
|---|---|---|
| Variable | `int unidades = 3;` | `unidades = 3` |
| Texto | `String nombre = "Cuaderno";` | `nombre = "Cuaderno"` |
| Imprimir | `System.out.println(nombre);` | `print(nombre)` |
| Condición | `if (unidades > 0) { ... }` | `if unidades > 0:` |
| Bucle | `for (int i=0; i<3; i++) { ... }` | `for i in range(3):` |

**Observa:** Python distingue mayúsculas y minúsculas; no exige punto y coma al terminar una instrucción. No es necesario declarar el tipo, aunque sigue siendo importante conocerlo.

````python
nombre = "Cuaderno"
precio = 2.50
unidades = 4
disponible = True

print("Producto:", nombre)
print("Tipos:", type(nombre).__name__, type(precio).__name__, type(unidades).__name__, type(disponible).__name__)
print("Valor de las unidades:", precio * unidades)
````

## 2. Listas, diccionarios y funciones

Una **lista** representa una secuencia ordenada de elementos; un **diccionario** asocia claves con valores. Para el catálogo de la tienda, cada producto puede describirse mediante un diccionario, y el catálogo completo puede ser una lista de diccionarios.

````python
catalogo = [
    {"codigo": "P001", "nombre": "Cuaderno", "precio": 2.50, "existencias": 20},
    {"codigo": "P002", "nombre": "Lápiz", "precio": 0.60, "existencias": 50},
    {"codigo": "P003", "nombre": "Mochila", "precio": 18.00, "existencias": 4},
]
````

````python
print("Primer producto:", catalogo[0]["nombre"])
for producto in catalogo:
    print(producto["codigo"], "-", producto["nombre"], "-", producto["existencias"], "unidades")
````

### Funciones

En Java, una operación similar se escribiría normalmente como un método dentro de una clase. En Python podemos comenzar con una función independiente usando `def` y `return`. Esta función recibe un producto y una cantidad y devuelve el subtotal, sin modificar el catálogo.

````python
def calcular_subtotal(precio, cantidad):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser positiva")
    return round(precio * cantidad, 2)


def buscar_producto(productos, codigo):
    for producto in productos:
        if producto["codigo"] == codigo:
            return producto
    return None
````

````python
producto = buscar_producto(catalogo, "P002")
print("Encontrado:", producto)
print("Subtotal por 3 lápices:", calcular_subtotal(producto["precio"], 3))
````

## 3. Programación orientada a objetos en Python

En Programación II trabajas con clases y objetos en Java. En Python se conservan los conceptos de **atributo**, **constructor**, **instancia** y **método**, pero cambia la sintaxis.

| Java | Python | Función |
|---|---|---|
| `class Producto { ... }` | `class Producto:` | Define una clase |
| `public Producto(...)` | `def __init__(self, ...):` | Inicializa el objeto |
| `this.precio` | `self.precio` | Accede al atributo del objeto |
| `new Producto(...)` | `Producto(...)` | Construye una instancia |

En Python, `self` es el parámetro que se refiere a la instancia; se escribe explícitamente en los métodos de instancia. `__init__` es un **inicializador** que se ejecuta al construir el objeto (en sentido estricto, no es el método que crea la instancia).

````python
class Producto:
    def __init__(self, codigo, nombre, precio, existencias):
        if precio < 0 or existencias < 0:
            raise ValueError("Precio y existencias no pueden ser negativos")
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.existencias = existencias

    def mostrar_info(self):
        return f"{self.codigo}: {self.nombre} | ${self.precio:.2f} | stock: {self.existencias}"

    def hay_existencias(self, cantidad):
        return cantidad > 0 and cantidad <= self.existencias

    def descontar_existencias(self, cantidad):
        if not self.hay_existencias(cantidad):
            raise ValueError("Cantidad inválida o existencias insuficientes")
        self.existencias -= cantidad
````

````python
cuaderno = Producto("P001", "Cuaderno", 2.50, 20)
print(cuaderno.mostrar_info())
print("¿Hay 3 unidades?", cuaderno.hay_existencias(3))
cuaderno.descontar_existencias(3)
print("Después de descontar:", cuaderno.mostrar_info())
````

### Del diccionario al objeto

Los diccionarios son útiles para transportar y guardar registros; los objetos también reúnen **comportamiento** (métodos). En esta práctica usaremos ambos: objetos para la lógica del producto y diccionarios para escribir CSV.

````python
productos = [
    Producto("P001", "Cuaderno", 2.50, 20),
    Producto("P002", "Lápiz", 0.60, 50),
    Producto("P003", "Mochila", 18.00, 4),
]
````

````python
for item in productos:
    print(item.mostrar_info())
````

## 4. Persistencia: Introducción a CSV y pandas

Los objetos creados en memoria se pierden al terminar el programa. Un archivo **CSV** permite guardar registros tabulares de forma sencilla. Cada fila representa un producto y cada columna un atributo.

**Importante:** CSV no es una base de datos ni garantiza por sí mismo transacciones atómicas. Hoy lo utilizaremos para persistencia básica.

````python
import csv
from pathlib import Path

ruta_csv = Path("productos.csv")
columnas = ["codigo", "nombre", "precio", "existencias"]

with ruta_csv.open("w", newline="", encoding="utf-8") as archivo:
    escritor = csv.DictWriter(archivo, fieldnames=columnas)
    escritor.writeheader()
    for item in productos:
        escritor.writerow({
            "codigo": item.codigo,
            "nombre": item.nombre,
            "precio": item.precio,
            "existencias": item.existencias,
        })

print("Archivo generado:", ruta_csv.resolve())
````

````python
with ruta_csv.open("r", newline="", encoding="utf-8") as archivo:
    lector = csv.DictReader(archivo)
    registros = list(lector)

print("Registros recuperados:", registros)
print("Tipo leído para precio:", type(registros[0]["precio"]).__name__)
print("Nota: csv.DictReader devuelve texto; convertir tipos antes de calcular.")
````

### pandas: una vista tabular del catálogo

`pandas` facilita la lectura y consulta de datos tabulares mediante el objeto `DataFrame`. En este notebook solo utilizaremos operaciones básicas: leer, seleccionar columnas y filtrar filas.

Si no tienes instalada la biblioteca en tu entorno, ejecuta en una celda separada: `%pip install pandas`.

````python
import pandas as pd

df_productos = pd.read_csv(ruta_csv, dtype={"codigo": "string"})
display(df_productos)

print("Productos con menos de 10 unidades:")
display(df_productos.loc[df_productos["existencias"] < 10, ["codigo", "nombre", "existencias"]])

print("Valor total del inventario:", round((df_productos["precio"] * df_productos["existencias"]).sum(), 2))
````
